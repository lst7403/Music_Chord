# Step 5: Beat & Chord Alignment (Measure Lead Sheet Engine)

This notebook uses **`pipeline.py`** to perform **beat-synchronous chord alignment**:
1. **Aligns** continuous AI chord intervals to discrete rhythm beats.
2. **Groups** beats into musical **Measures (Bars)** based on detected downbeats (4/4 time).
3. **Quantizes** harmonic rhythm into standard **Lead Sheet** sheet music format.
4. **Exports** results to **`data/aligned_chords.csv`** for the web application.

> **Kernel**: Make sure the kernel is set to **`Python (seperate)`**.

## 1. Setup & Imports

```python
from pathlib import Path
import pandas as pd
import numpy as np
from pipeline import align_chords_to_beats, save_aligned_chords

data_dir = Path("data")
chords_path = data_dir / "chords.csv"
beats_path = data_dir / "beats.csv"
aligned_path = data_dir / "aligned_chords.csv"

assert chords_path.exists(), f"Missing chords file: {chords_path}. Run chord_recognition.ipynb first!"
assert beats_path.exists(), f"Missing beats file: {beats_path}. Run beat.ipynb first!"

print(f"✓ Chords input : {chords_path}")
print(f"✓ Beats input  : {beats_path}")
print(f"✓ Target output: {aligned_path}")
```

## 2. Run Beat-Synchronous Chord Alignment

```python
results = align_chords_to_beats(chords_path, beats_path)
df_aligned = results["df"]

print(f"✅ Successfully aligned {results['total_beats']} beats into {results['total_measures']} musical measures!")
print(f"Tempo: {results['bpm']} BPM")

print("\nFirst 16 beat-chord records:")
print(df_aligned.head(16).to_string(index=False))
```

## 3. Save Aligned Chords to CSV

```python
save_aligned_chords(results, aligned_path)
print(f"✅ Saved beat-aligned chord sheet to: {aligned_path}")
```

## 4. Measure-by-Measure Lead Sheet Breakdown

```python
print("🎵 Lead Sheet Measure Breakdown (First 15 Bars):")
print("-" * 65)
print(f"{'BAR':<6} | {'TIME RANGE':<16} | {'BEATS (CHORD PER BEAT)':<25} | {'CHORD SUMMARY'}")
print("-" * 65)

for m in results["measures"][:15]:
    bar_str = f"Bar {m['measure']}"
    time_str = f"{m['start']:.2f}s - {m['end']:.2f}s"
    beat_chords = " | ".join(f"b{b['beat']}:{b['chord']}" for b in m["beats"])
    print(f"{bar_str:<6} | {time_str:<16} | {beat_chords:<25} | {m['summary']}")

print("-" * 65)
```
