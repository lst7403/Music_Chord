# Step 2: Combine Audio Stems into Backing & Harmonic Tracks

This notebook uses **`module.source_separation`** to combine isolated audio stems from `separate.ipynb`:
1. **Harmonic Accompaniment** (`data/bass_other.wav` = `bass.wav` + `other.wav`): Clean harmonic signal for BTC chord recognition.
2. **Full Instrumental Backing Track** (`data/instrumental.wav` = `drums.wav` + `bass.wav` + `other.wav`): Full accompaniment with rhythm section for the web app.

> **Kernel**: Make sure the kernel is set to **`Python (seperate)`**.

## 1. Setup & Imports from `module.source_separation`

```python
import os
import sys
from pathlib import Path

# Ensure project root is in sys.path and set as current working directory
PROJECT_ROOT = Path("..").resolve() if Path("..", "module").exists() else Path(".").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

from pathlib import Path
from module.source_separation import combine_bass_and_other, combine_instrumental

data_dir = Path("data")
vocals_path = data_dir / "vocals.wav"
drums_path = data_dir / "drums.wav"
bass_path = data_dir / "bass.wav"
other_path = data_dir / "other.wav"

for p in [drums_path, bass_path, other_path]:
    assert p.exists(), f"Missing stem: {p}. Please run separate.ipynb first!"

print("✓ Required stem files (drums, bass, other) found in data/!")
```

## 2. Generate Harmonic Accompaniment (`data/bass_other.wav`)
Merges `bass.wav` + `other.wav` for chord recognition.

```python
accompaniment_path = data_dir / "bass_other.wav"

combine_bass_and_other(bass_path, other_path, accompaniment_path)

size_mb = accompaniment_path.stat().st_size / (1024 * 1024)
print(f"✅ Successfully created Harmonic Accompaniment: {accompaniment_path} ({size_mb:.2f} MB)")
```

## 3. Generate Instrumental Backing Track (`data/instrumental.wav`)
Merges `drums.wav` + `bass.wav` + `other.wav` for full backing track playback.

```python
instrumental_path = data_dir / "instrumental.wav"

combine_instrumental(drums_path, bass_path, other_path, instrumental_path)

size_mb = instrumental_path.stat().st_size / (1024 * 1024)
print(f"✅ Successfully created Instrumental Backing Track: {instrumental_path} ({size_mb:.2f} MB)")
```
